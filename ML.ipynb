```python
def filter_even_numbers(numbers):
  result = []
  for w in numbers  :
    if (w % 2 == 0)  :
      result.append(w)
  return result
```

```python
ll = [1 ,3 ,4, 6, 8, 7 ,5 ]
filter_even_numbers(ll)
```
